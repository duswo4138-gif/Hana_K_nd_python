```python
def average(*a):
    result = 0

    for i in a:
        result = result + i  # 누적합

    avg = result / len(a) # 3으로 나눠주면 평균

    return avg

average(10,20,30)
```

```python

# 두 숫자 입력받기
num1 = float(input("첫 번째 숫자를 입력하세요: "))
num2 = float(input("두 번째 숫자를 입력하세요: "))

# 연산자 입력받기
operator = input("연산자를 입력하세요 (+, -, *, /): ")

# 연산 수행
if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "0으로 나눌 수 없습니다."
else:
    result = "잘못된 연산자입니다."

# 결과 출력
print("계산 결과:", result)
```

```python

# 설치: pip install yfinance pandas

import yfinance as yf
import pandas as pd

# 1. 투자 조건
ticker = "NVDA"
investment_krw = 10_000_000

start_date = "2020-01-02"
end_date = pd.Timestamp.today().strftime("%Y-%m-%d")

# 2. 주가 데이터 다운로드
# auto_adjust=True: 주식분할 및 배당 조정 반영
stock = yf.download(
    ticker,
    start=start_date,
    end=end_date,
    auto_adjust=True,
    progress=False
)

# 3. 원/달러 환율 데이터 다운로드
fx = yf.download(
    "KRW=X",
    start=start_date,
    end=end_date,
    auto_adjust=True,
    progress=False
)

# 4. 데이터 정리
stock = stock.dropna(subset=["Close"])
fx = fx.dropna(subset=["Close"])

buy_price = float(stock["Close"].iloc[0].item())
current_price = float(stock["Close"].iloc[-1].item())

buy_fx = float(fx["Close"].iloc[0].item())
current_fx = float(fx["Close"].iloc[-1].item())

# 5. 투자금 원화 -> 달러 환전
investment_usd = investment_krw / buy_fx

# 매수 가능한 소수점 주식까지 보유한다고 가정
shares = investment_usd / buy_price

# 6. 현재 평가금액 계산
current_value_usd = shares * current_price
current_value_krw = current_value_usd * current_fx

# 7. 수익 계산
profit_krw = current_value_krw - investment_krw
return_rate = (profit_krw / investment_krw) * 100

# 8. 결과 출력
print("=" * 40)
print("엔비디아(NVDA) 투자 수익 분석")
print("=" * 40)

print(f"매수일 주가: ${buy_price:,.2f}")
print(f"최근 종가: ${current_price:,.2f}")
print(f"매수 시 환율: {buy_fx:,.2f}원")
print(f"최근 환율: {current_fx:,.2f}원")
print(f"보유 주식 수: {shares:,.4f}주")
print(f"초기 투자금: {investment_krw:,.0f}원")
print(f"현재 평가금액: {current_value_krw:,.0f}원")
print(f"총 수익금: {profit_krw:,.0f}원")
print(f"수익률: {return_rate:,.2f}%")
print("=" * 40)

```

```python
import yfinance as yf

# 투자금
investment = 10_000_000 # 1,000만원

# 엔비디아 2020년 주가 데이터
nvda = yf.download(
    "NVDA",
    start="2020-01-01",
    end="2021-01-01",
    auto_adjust=True
)

# 2020년 첫 거래일 / 마지막 거래일 종가
start_price = float(nvda["Close"].iloc[0])
end_price = float(nvda["Close"].iloc[-1])

# 주가 상승률
return_rate = (end_price - start_price) / start_price

# 최종 금액
final_money = investment * (1 + return_rate)

# 수익금
profit = final_money - investment

print(f"2020년 시작 주가: ${start_price:.2f}")
print(f"2020년 마지막 주가: ${end_price:.2f}")
print(f"수익률: {return_rate * 100:.2f}%")
print(f"투자금: {investment:,.0f}원")
print(f"최종 금액: {final_money:,.0f}원")
print(f"수익금: {profit:,.0f}원")
```
